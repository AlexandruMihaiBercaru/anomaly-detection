```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
sample_size = 200

X = np.random.normal(size=sample_size)
a, b = 2 , -3
Y = a * X + b
```

```python
# low noise => mean close to 0, standard deviation close to 1
mu = 0.5
sigma = 1.5
epsilon = np.random.normal(loc=mu, scale = sigma, size=sample_size)

Y = a * X + b + epsilon
print(X.shape)
print(Y.shape)
```

```python
plt.scatter(X, Y, c='red', marker='o', s=1)
plt.xlabel("Feature X")
plt.ylabel("Label Y")
plt.show()
```

```python
# compute the hat matrix
X = X.reshape(-1,1)
H = X @ np.linalg.inv(X.T @ X) @ X.T
print(H.shape)

# extract leverage scores
leverage_scores = np.diagonal(H)
print(leverage_scores.shape)

# sanity check
sum_lev = np.sum(leverage_scores)
print(sum_lev) # for 1D case, should be 1

ind = np.argsort(leverage_scores)
top_k = ind[-16:]
print(top_k)

X_high = X[top_k]
Y_high = Y[top_k]
print(X_high)

print(X_high.shape)
print(Y_high.shape)
plt.plot(X, a*X+b, color='black', linewidth=1)
plt.scatter(X, Y, c='red', marker='o', s=1)
plt.scatter(X_high, Y_high, c='blue', s=7)
plt.xlabel("Feature X")
plt.ylabel("Label Y")
plt.show()
```
